# Data Understanding

* Dataset: `Zomato.csv`
* Shape: 9,551 rows × 21 columns
* Each row represents a restaurant.
* `Restaurant ID` is unique; no duplicate rows.
* Only 9 missing values in `Cuisines`.
* 15 countries and 41 cities are represented.
* India is the dominant market with **8,652 restaurants**.
* ,825 unique cuisine combinations.
* Price Range 1 has the highest number of restaurants.
* Average rating: 2.67; median: 3.2.
* 2,148 restaurants are Not Rated.
* Median votes: 31; maximum: 10,934.
* Online delivery: 2,451 restaurants; table booking: 1,158 restaurants.
* `Country-Code.xlsx` provides the country-name mapping.
