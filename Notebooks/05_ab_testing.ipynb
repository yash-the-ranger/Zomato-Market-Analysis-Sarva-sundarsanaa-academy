```python
import pandas as pd
from scipy import stats

df = pd.read_csv(
    r"D:\my career\Zomato market analysis\Datasets\processed\zomato_featured.csv"
)

df.shape
```

Create the two groups

```python
online_delivery_yes = df.loc[
    df["Has Online delivery"] == "Yes", "Votes"
]

online_delivery_no = df.loc[
    df["Has Online delivery"] == "No", "Votes"
]

online_delivery_yes.shape, online_delivery_no.shape
```

Calculate group means

```python
online_delivery_yes.mean(), online_delivery_no.mean()
```

Statistical test

```python
t_stat, p_value = stats.ttest_ind(
    online_delivery_yes,
    online_delivery_no,
    equal_var=False
)

t_stat, p_value
```

Online Delivery vs Votes

```python
test_results = pd.DataFrame({
    "Test": ["Online Delivery vs Votes"],
    "Group 1": ["Online Delivery = Yes"],
    "Group 2": ["Online Delivery = No"],
    "Mean 1": [online_delivery_yes.mean()],
    "Mean 2": [online_delivery_no.mean()],
    "t_statistic": [t_stat],
    "p_value": [p_value],
    "Significant": [p_value < 0.05]
})

test_results
```

Table Booking vs Rating

```python
table_booking_yes = df.loc[
    df["Has Table booking"] == "Yes", "Aggregate rating"
]

table_booking_no = df.loc[
    df["Has Table booking"] == "No", "Aggregate rating"
]

table_booking_yes.shape, table_booking_no.shape
```

group means

```python
table_booking_yes.mean(), table_booking_no.mean()
```

two-sample t-test

```python
t_stat_2, p_value_2 = stats.ttest_ind(
    table_booking_yes,
    table_booking_no,
    equal_var=False
)

t_stat_2, p_value_2
```

Table Booking vs Rating

```python
test_results.loc[len(test_results)] = {
    "Test": "Table Booking vs Rating",
    "Group 1": "Table Booking = Yes",
    "Group 2": "Table Booking = No",
    "Mean 1": table_booking_yes.mean(),
    "Mean 2": table_booking_no.mean(),
    "t_statistic": t_stat_2,
    "p_value": p_value_2,
    "Significant": p_value_2 < 0.05
}

test_results
```
